```python
import zipfile
import os

# Paths to your zip files
fake_zip_path = "/content/fake_cifake_images-20251019T111046Z-1-001.zip"
real_zip_path = "/content/real_cifake_images-20251019T111045Z-1-001.zip"
test_zip_path = "/content/test-20251019T111049Z-1-001.zip"

# Destination folders
fake_dest = "/content/fake_images"
real_dest = "/content/real_images"
test_dest = "/content/test_images"

# Function to extract all files from zip into a destination folder
def extract_zip(zip_path, dest_folder):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(dest_folder)
    print(f"Extracted {zip_path} to {dest_folder}")

# Create folders if they don't exist
os.makedirs(fake_dest, exist_ok=True)
os.makedirs(real_dest, exist_ok=True)
os.makedirs(test_dest, exist_ok=True)

# Extract all
extract_zip(fake_zip_path, fake_dest)
extract_zip(real_zip_path, real_dest)
extract_zip(test_zip_path, test_dest)

# Count images
print("Fake images:", len(os.listdir(os.path.join(fake_dest, "fake_cifake_images"))))
print("Real images:", len(os.listdir(os.path.join(real_dest, "real_cifake_images"))))
print("Test images:", len(os.listdir(os.path.join(test_dest, "test"))))
```

```python
import os
from sklearn.model_selection import train_test_split

# Paths to extracted folders
fake_folder = "/content/fake_images/fake_cifake_images"
real_folder = "/content/real_images/real_cifake_images"
test_folder = "/content/test_images/test"

# Create lists of file paths with labels (1 for fake, 0 for real)
fake_images = [(os.path.join(fake_folder, f), 1) for f in os.listdir(fake_folder)]
real_images = [(os.path.join(real_folder, f), 0) for f in os.listdir(real_folder)]

# Combine fake and real
all_images = fake_images + real_images

# Split into training and validation sets (80% train, 20% val)
train_data, val_data = train_test_split(all_images, test_size=0.2, random_state=42, shuffle=True)

print("Training images:", len(train_data))
print("Validation images:", len(val_data))
print("Test images:", len(os.listdir(test_folder)))
```

```python
import pandas as pd

# Separate filenames and labels
train_images = [i[0] for i in train_data]
train_labels = [i[1] for i in train_data]
val_images = [i[0] for i in val_data]
val_labels = [i[1] for i in val_data]

# Convert labels to strings (required by ImageDataGenerator)
train_df = pd.DataFrame({
    'filename': train_images,
    'class': [str(x) for x in train_labels]
})

val_df = pd.DataFrame({
    'filename': val_images,
    'class': [str(x) for x in val_labels]
})

print("✅ DataFrames ready!")
train_df.head()
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Data augmentation for training
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2
)

# Only rescaling for validation
val_datagen = ImageDataGenerator(rescale=1./255)

# Create data generators
train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col='filename',
    y_col='class',
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary'
)

validation_generator = val_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col='filename',
    y_col='class',
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary'
)
```

```python
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras import layers, models, optimizers

# Base pre-trained model
base_model = EfficientNetB3(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Freeze base model (transfer learning)
base_model.trainable = False

# Add custom layers
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.4),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')  # Binary classification
])

# Compile model
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
# Train the model
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=15,
    verbose=1
)
```

```python
# Unfreeze top layers of EfficientNetB3
base_model.trainable = True

# Freeze bottom layers (keep low-level features)
for layer in base_model.layers[:-50]:  # unfreeze last 50 layers
    layer.trainable = False

# Recompile with a lower learning rate for fine-tuning
from tensorflow.keras import optimizers

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),  # lower LR for fine-tuning
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Stop training if val_loss doesn't improve for 5 epochs
early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)

# Save the best model during training
checkpoint = ModelCheckpoint(
    'best_deepfake_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    verbose=1
)
```

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20,          # can increase later if needed
    batch_size=32,
    callbacks=[early_stop, checkpoint],
    verbose=1
)
```

```python
import os
import json
import numpy as np
import pandas as pd
from mtcnn import MTCNN
from PIL import Image
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
```

```python
!pip install lz4
from mtcnn import MTCNN
detector = MTCNN()

detector = MTCNN()

def crop_face(image_path, target_size=(224,224)):
    img = Image.open(image_path).convert('RGB')
    pixels = np.array(img)
    faces = detector.detect_faces(pixels)
    if faces:
        x, y, w, h = faces[0]['box']
        face = pixels[y:y+h, x:x+w]
        face_img = Image.fromarray(face).resize(target_size)
        return np.array(face_img)
    else:
        # If no face detected, return resized original
        return np.array(img.resize(target_size))
```

```python
# Paths
import os
fake_folder = "/content/fake_images/fake_cifake_images"
real_folder = "/content/real_images/real_cifake_images"
test_folder = "/content/test_images/test"

# Create lists of file paths with labels
fake_images = [(os.path.join(fake_folder, f), 1) for f in os.listdir(fake_folder)]
real_images = [(os.path.join(real_folder, f), 0) for f in os.listdir(real_folder)]
all_images = fake_images + real_images

# Split train/val
train_data, val_data = train_test_split(all_images, test_size=0.2, random_state=42, shuffle=True)

# Convert to dataframe for ImageDataGenerator
train_df = pd.DataFrame({
    'filename':[i[0] for i in train_data],
    'class':[str(i[1]) for i in train_data]
})

val_df = pd.DataFrame({
    'filename':[i[0] for i in val_data],
    'class':[str(i[1]) for i in val_data]
})
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    zoom_range=0.3,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
    brightness_range=[0.7,1.3],
    shear_range=0.2
)

val_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_dataframe(
    train_df,
    x_col='filename',
    y_col='class',
    target_size=(224,224),
    batch_size=32,
    class_mode='binary',
    shuffle=True
)

validation_generator = val_datagen.flow_from_dataframe(
    val_df,
    x_col='filename',
    y_col='class',
    target_size=(224,224),
    batch_size=32,
    class_mode='binary'
)
```

```python
base_model = EfficientNetB3(
    input_shape=(224,224,3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False  # start with frozen layers

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.4),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model.summary()
```

```python
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)
checkpoint = ModelCheckpoint('best_deepfake_model.h5', monitor='val_accuracy', save_best_only=True, verbose=1)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1)
```

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)
```

```python
# Unfreeze top layers for fine-tuning
base_model.trainable = True
for layer in base_model.layers[:-50]:  # freeze bottom layers
    layer.trainable = False

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history_finetune = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=15,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)
```

```python
import json
import os
import numpy as np

test_images = sorted(os.listdir(test_folder))  # ensure order matches "index"
predictions = []

for idx, img_name in enumerate(test_images, start=1):
    img_path = os.path.join(test_folder, img_name)
    img = crop_face(img_path)
    img = np.expand_dims(img / 255., axis=0)
    pred = model.predict(img)[0][0]
    label = "fake" if pred > 0.5 else "real"  # threshold 0.5
    predictions.append({"index": idx, "prediction": label})

with open('Nmaangers.json', 'w') as f:
    json.dump(predictions, f, indent=4)

print("✅ Predictions saved in hackathon format to Nmaangers.json")
```

```python
!pip install mtcnn lz4

import os
import json
import numpy as np
import pandas as pd
from PIL import Image
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from mtcnn import MTCNN
```

```python
detector = MTCNN()

def crop_face(image_path, target_size=(224,224)):
    img = Image.open(image_path).convert('RGB')
    pixels = np.array(img)
    faces = detector.detect_faces(pixels)
    if faces:
        x, y, w, h = faces[0]['box']
        # Ensure positive coordinates
        x, y = max(0,x), max(0,y)
        face = pixels[y:y+h, x:x+w]
        face_img = Image.fromarray(face).resize(target_size)
        return np.array(face_img)
    else:
        # If no face detected, return resized original
        return np.array(img.resize(target_size))
```

```python
fake_folder = "/content/fake_images/fake_cifake_images"
real_folder = "/content/real_images/real_cifake_images"
test_folder = "/content/test_images/test"

fake_images = [(os.path.join(fake_folder, f), 1) for f in os.listdir(fake_folder)]
real_images = [(os.path.join(real_folder, f), 0) for f in os.listdir(real_folder)]
all_images = fake_images + real_images

train_data, val_data = train_test_split(all_images, test_size=0.2, random_state=42, shuffle=True)

train_df = pd.DataFrame({
    'filename': [i[0] for i in train_data],
    'class': [str(i[1]) for i in train_data]
})

val_df = pd.DataFrame({
    'filename': [i[0] for i in val_data],
    'class': [str(i[1]) for i in val_data]
})

print("✅ Training:", len(train_df), "Validation:", len(val_df))
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    zoom_range=0.3,
    horizontal_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
    brightness_range=[0.7,1.3],
    shear_range=0.2
)

val_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_dataframe(
    train_df, x_col='filename', y_col='class',
    target_size=(224,224), batch_size=32,
    class_mode='binary', shuffle=True
)

validation_generator = val_datagen.flow_from_dataframe(
    val_df, x_col='filename', y_col='class',
    target_size=(224,224), batch_size=32,
    class_mode='binary'
)
```

```python
base_model = EfficientNetB3(input_shape=(224,224,3), include_top=False, weights='imagenet')
base_model.trainable = False  # freeze initially

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.4),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer=optimizers.Adam(learning_rate=1e-4),
              loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)
checkpoint = ModelCheckpoint('best_deepfake_model.h5', monitor='val_accuracy', save_best_only=True, verbose=1)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1)
```

```python
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=20,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)
```

```python
base_model.trainable = True
for layer in base_model.layers[:-150]:
    layer.trainable = False

model.compile(optimizer=optimizers.Adam(1e-5),
              loss='binary_crossentropy', metrics=['accuracy'])

history_finetune = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=15,
    callbacks=[early_stop, checkpoint, reduce_lr],
    verbose=1
)
```

```python
test_images = [os.path.join(test_folder, f) for f in os.listdir(test_folder)]
predictions = []

for idx, img_path in enumerate(test_images, 1):
    img = crop_face(img_path)
    img = np.expand_dims(img/255., axis=0)
    pred = model.predict(img)[0][0]
    label = "real" if pred < 0.5 else "fake"
    predictions.append({"index": idx, "prediction": label})

with open('Nmaangers.json', 'w') as f:
    json.dump(predictions, f, indent=4)

print("✅ Predictions saved to Nmaangers.json")
```
